# HW2: Publish a Web Page with GitHub Pages

Complete HW2 in your own copy of the [HW2 website template repository](https://github.com/macss-berkeley/hw2-hidden-berkeley-template). It contains the assignment notebook, CSV, and website starter. This course notebook is a guide to getting started; record your answers in the template's notebook.

## Open the right repository

1. Complete the [course setup](../SETUP.md) and sign into GitHub.com and GitHub Desktop with the same account.
2. Open the template and choose **Use this template → Create a new repository**. Create a public repository under your account. Use `hidden-berkeley`, or `hidden-berkeley-hw2` if the first name is already taken.
3. In GitHub Desktop, choose **File → Clone Repository → URL** and paste **your new repository's URL**. Save it outside the course repository, then open its complete folder in VS Code.
4. Open `hw2_hidden_berkeley.ipynb` there. Select the course Python environment as its kernel. If needed, browse to the course repository's `.venv/bin/python` (macOS/Linux) or `.venv\Scripts\python.exe` (Windows).

You should see the notebook, `README.md`, and `docs/` in the same folder. The website reads `docs/_data/locations.csv`; edit that CSV through the assignment notebook. Edit the page heading and introduction in `docs/index.md` as instructed, retaining the supplied configuration and layout.

## Read or watch the Pages walkthrough

The `/docs` folder contains the supplied site. The blog uses a different example: skip its sample-repository fork, Next.js, custom GitHub Actions workflow, and custom-domain sections. You do not need those for HW2. The **Actions** tab still shows GitHub's automatic Pages build; viewing it does not require writing a workflow.

Wait for the build to finish and open the website address shown in Pages settings. Before editing data, check that the eight starter resources appear. After your changes, the assignment asks for nine resources, including one improved entry and one new entry.

## Follow one change through the workflow

| Step | Where the change lives | What to check |
| --- | --- | --- |
| Run a pandas edit | In the kernel's temporary DataFrame | Display the changed row. |
| Run the supplied CSV save cell | In the CSV on your computer | Read it back; keep the `location_id` index as instructed. |
| Save the notebook | In the `.ipynb` file on your computer | Your code and outputs are saved. This alone does not save the CSV. |
| Commit in GitHub Desktop | In local Git history | Inspect the diff, then record the change with a descriptive message. |
| Push origin | In your GitHub repository | Open the CSV on GitHub and find your edit. |
| Pages build finishes | On the published website | Open the site and verify the displayed entry and link. |

A **repository** stores files and their history. A **commit** is a saved version in that history; **push** sends local commits to GitHub. Your repository URL shows code (`github.com/...`); your Pages URL shows the website (`...github.io/...`). Submit both actual URLs, not these examples.

## If something is missing

- **No CSV found:** check that VS Code opened your HW2 repository and that `docs/_data/locations.csv` is present. Selecting the course kernel does not change the working folder.
- **No change in Desktop:** confirm the selected repository, save the notebook, and run the CSV save cell if you changed data.
- **GitHub has the old CSV:** check for an unpushed commit in Desktop.
- **GitHub has the new CSV but the website is old or returns 404:** confirm `main` and `/docs`, inspect the latest build in **Actions**, then refresh after it succeeds. Use the URL from Pages settings.
- **No Pages settings:** confirm that you are in the repository under your own account, not the course or template repository.
