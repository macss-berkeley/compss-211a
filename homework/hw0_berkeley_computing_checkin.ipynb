```python
from pathlib import Path
import json
import os
import sys

import importlib.metadata as metadata
import platform
import pandas as pd
from IPython.display import display

SUPPORTED_PYTHON = (3, 13)
if sys.version_info[:2] != SUPPORTED_PYTHON:
    print(
        "Setup note: this course is tested with Python 3.13; "
        f"you are running {sys.version.split()[0]}."
    )

def find_course_root():
    """Find the cloned repository when this notebook is running locally."""
    for folder in (Path.cwd(), *Path.cwd().parents):
        if (folder / "pyproject.toml").exists() and (folder / "data").is_dir():
            return folder
    return None

LOCAL_COURSE_ROOT = find_course_root()
COURSE_ROOT = LOCAL_COURSE_ROOT or Path.cwd()
DATA_DIR = (
    LOCAL_COURSE_ROOT / "data"
    if LOCAL_COURSE_ROOT
    else COURSE_ROOT / "compss211_data"
)
GENERATED_DIR = COURSE_ROOT / "generated"
DATA_DIR.mkdir(parents=True, exist_ok=True)
GENERATED_DIR.mkdir(parents=True, exist_ok=True)

DATA_BASE_URL = (
    "https://raw.githubusercontent.com/"
    "macss-berkeley/compss-211a/main/data"
)

def course_data_path(filename):
    """Use a local course file, or download it when running in Colab."""
    path = DATA_DIR / filename
    if not path.exists():
        from urllib.request import urlretrieve

        urlretrieve(f"{DATA_BASE_URL}/{filename}", path)
    return path

print(f"Python {sys.version.split()[0]} | data={DATA_DIR}")
```

## 1. Inspect the environment

```python
package_names = ["pandas", "numpy", "matplotlib", "scikit-learn", "requests"]
environment_summary = {
    "python_version": sys.version.split()[0],
    "platform": platform.platform(),
    "working_directory": str(Path.cwd()),
    "course_root": str(COURSE_ROOT),
    "local_clone_found": LOCAL_COURSE_ROOT is not None,
    "uv_lock_found": bool(
        LOCAL_COURSE_ROOT and (LOCAL_COURSE_ROOT / "uv.lock").exists()
    ),
    "project_environment": Path(sys.prefix).name == ".venv",
}
package_versions = {name: metadata.version(name) for name in package_names}
display(pd.Series(environment_summary, name="value"))
display(pd.Series(package_versions, name="version"))
```

## 2. Run the Pandas smoke test

```python
recap = pd.read_csv(course_data_path("hw1_after_dark_commute_survey.csv"))
recap_summary = (
    recap.groupby("commute_mode", dropna=False)
    .agg(respondents=("respondent_id", "count"))
    .sort_values("respondents", ascending=False)
)
recap_summary.to_csv(GENERATED_DIR / "hw0_commute_mode_counts.csv")
display(recap.head(3))
display(recap_summary)
```
