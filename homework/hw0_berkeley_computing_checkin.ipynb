## 2. Run the Pandas smoke test

```python
recap = pd.read_csv(DATA_DIR / "hw1_after_dark_commute_survey.csv")
recap_summary = (
    recap.groupby("commute_mode", dropna=False)
    .agg(respondents=("respondent_id", "count"))
    .sort_values("respondents", ascending=False)
)
recap_summary.to_csv(GENERATED_DIR / "hw0_commute_mode_counts.csv")
display(recap.head(3))
display(recap_summary)
```
