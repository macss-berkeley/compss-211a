## Practice: generate one accessible field-guide card

```python
import html
locations = pd.read_csv(DATA_DIR / "hw2_hidden_berkeley_locations.csv")

def render_card(row):
    return (
        f"<article><h2>{html.escape(row['name'])}</h2>"
        f"<p>{html.escape(row['access_note'])}</p>"
        f"<a href='{html.escape(row['source_url'], quote=True)}'>"
        "Official source</a></article>"
    )

card_html = render_card(locations.iloc[0])
page_path = GENERATED_DIR / "lab03_index.html"
page_path.write_text(card_html, encoding="utf-8")
display(card_html)
```
