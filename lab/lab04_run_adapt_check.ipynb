<details>
<summary>Reference: the function practised on Monday</summary>

```python
def category_report(requests, category):
    selected = requests.loc[requests["category"] == category].copy()
    report = summarize_requests(selected, group_by="district")
    report["pct_with_time"] = (
        100 * report["with_time"] / report["requests"]
    ).round(1)
    return report
```

The function selects one category, reuses the district summary, and adds the percentage of requests with usable durations. It returns a table; it does not save a CSV.
