## Practice: text cleaning and TF-IDF

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
comments = pd.read_csv(DATA_DIR / "hw4_synthetic_campus_comments.csv")

def clean_comment(text):
    cleaned = re.sub(r"[^a-z0-9\s'-]", " ", str(text).lower())
    return re.sub(r"\s+", " ", cleaned).strip()

def rank_terms(vectorizer, matrix, n=12):
    weights = np.asarray(matrix.mean(axis=0)).ravel()
    return pd.DataFrame({
        "term": vectorizer.get_feature_names_out(),
        "weight": weights,
    }).nlargest(n, "weight")

comments["clean_text"] = comments["text"].map(clean_comment)
vectorizer = TfidfVectorizer(stop_words="english", ngram_range=(1, 2), min_df=2)
matrix = vectorizer.fit_transform(comments["clean_text"])
top = rank_terms(vectorizer, matrix)
display(top)
```
