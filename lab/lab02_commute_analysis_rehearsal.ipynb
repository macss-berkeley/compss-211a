## Practice: a fresh commute question

```python
commute = pd.read_csv(course_data_path("hw1_after_dark_commute_survey.csv"))

def summarize_late_commute(frame):
    rehearsal = frame.loc[
        frame["commute_mode"].ne("Remote")
    ].copy()
    rehearsal["late_share"] = (
        rehearsal["leaves_after_8pm_days"]
        / rehearsal["days_on_campus"]
    )
    summary = rehearsal.groupby("commute_mode").agg(
        respondents=("respondent_id", "count"),
        median_late_share=("late_share", "median"),
    ).sort_values("median_late_share", ascending=False)
    return rehearsal, summary

rehearsal, late_summary = summarize_late_commute(commute)
display(late_summary)
```
