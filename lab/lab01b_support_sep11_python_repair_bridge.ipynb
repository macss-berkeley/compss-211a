## 2. Read an error and find the problem

Imagine that we tried to select a column named `seat` and Python showed this message:

```text
KeyError: 'seat'
```

Start with the **last line**. `KeyError: 'seat'` means Python could not find the label `seat`. Save that label so we can check it.

````python
wrong_column = "seat"
````

Ask whether that label is in the table’s column names. Predict `True` or `False` before running it.

````python
wrong_column in tiny.columns
````

Compare the requested label with the available names. The table has `seats`, with an **s**.

````python
correct_column = "seats"
````

Select the repaired column and save it as `seat_counts`.

````python
seat_counts = tiny[correct_column]
````

Show the values in the repaired column.

````python
seat_counts
````

## 3. Notice the missing value

`NaN` means a value is missing. It does not mean zero. Ask which place has a missing seat count.

````python
seat_counts.isna()
````

Count how many seat values are missing. In Python, `True` counts as 1.

````python
seat_counts.isna().sum()
````

Keep only the seat counts that are present.

````python
observed_seats = seat_counts.dropna()
````

Show the observed values.

````python
observed_seats
````

Find the average of the observed seat counts.

````python
observed_average = observed_seats.mean()
````

Show the observed-value average.

````python
observed_average
````

## 4. Compare a different missing-value choice

Here we temporarily replace the missing seat count with zero. This asks a different question because an unknown count becomes a known zero.

````python
zero_filled_seats = seat_counts.fillna(0)
````

Show the values after filling the missing value.

````python
zero_filled_seats
````

Find the average after replacing the missing value with zero.

````python
zero_filled_average = zero_filled_seats.mean()
````

Show the new average. Is it the same as the observed-value average?

````python
zero_filled_average
````

## 5. Choose rows after the repair

First, remove rows where the seat count is missing. This gives us a table with observed seat counts.

````python
known_seats = tiny.dropna(subset=["seats"])
````

Make a rule that asks which observed counts are at least 20.

````python
has_20_seats = known_seats["seats"] >= 20
````

Use the rule to keep those rows.

````python
larger_places = known_seats[has_20_seats]
````

Show the result.

````python
larger_places
````

## 6. Make and use a function

A function gives a short name to a reusable action. `count` is the input. The `return` line sends a result back.

````python
def seat_message(count):
    return f"{count} seats available"
````

Call the function with the number 18. The value inside the parentheses becomes `count`.

````python
seat_message(18)
````

Call the same function with our observed-value average.

````python
seat_message(observed_average)
````
