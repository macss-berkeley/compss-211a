## 1. Open the table

We have loaded pandas as `pd`. Now use it to open our practice table.

Read the CSV file and save the table with the name `commute`.

```python
commute = pd.read_csv(data_url)
```

Show the first five rows. Each row represents one person in our made-up survey.

```python
commute.head()
```

Show the size of the table. The first number is the number of rows. The second is the number of columns.

```python
commute.shape
```

## 2. Look at one column

`leaves_after_8pm_days` tells us how many days each person leaves campus after 8 p.m.

Select that column and save it as `late_days`.

```python
late_days = commute["leaves_after_8pm_days"]
```

Show the first five values in that column.

```python
late_days.head()
```

## 3. Choose people who leave late at least twice

`>= 2` means **2 or more**. This line asks whether each value is at least 2 and saves the answers.

```python
leaves_late = late_days >= 2
```

Show the first five answers. `True` means the row meets our rule. `False` means it does not.

```python
leaves_late.head()
```

Keep the rows where the answer is `True`. Save this smaller table as `evening`.

```python
evening = commute[leaves_late]
```

Look at the first five rows of the smaller table.

```python
evening.head()
```

How many rows did we keep? Read the first number below.

```python
evening.shape
```

## 4. Count commute types

Select the commute type for the people in our smaller table.

```python
evening_modes = evening["commute_mode"]
```

Count how often each commute type appears. Save the result as `mode_counts`.

```python
mode_counts = evening_modes.value_counts()
```

Show the counts. A larger count means more people in this smaller table use that commute type.

```python
mode_counts
```

## 5. Try a different rule

What if we keep only people who leave late **3 or more days**? Before you run the next cells, predict whether we will keep more or fewer rows.

```python
three_days = late_days >= 3
```

Use the new rule to choose rows from the original table.

```python
three_day_commutes = commute[three_days]
```

Show the size of the new table. Was your prediction right?

```python
three_day_commutes.shape
```

## 6. Show just two columns

A smaller table can be easier to read. Make a list of the two column names we want to see.

```python
column_names = ["commute_mode", "leaves_after_8pm_days"]
```

Use that list to select the columns. Save the result as `small_table`.

```python
small_table = commute[column_names]
```

Look at the first five rows. How many columns are shown now?

```python
small_table.head()
```

## 7. Put the largest values first

Sort the table by the number of late days. `ascending=False` means largest to smallest. Save this order as `sorted_commutes`.

```python
sorted_commutes = commute.sort_values("leaves_after_8pm_days", ascending=False)
```

Show the first five rows in this order. What is the largest number of late days?

```python
sorted_commutes.head()
```

## 8. Find an average

An average adds the values and divides by how many values there are. Use `mean()` on our `late_days` column and save the answer.

```python
average_late_days = late_days.mean()
```

Show the average number of late days per person in this made-up table.

```python
average_late_days
```

Now show the largest value with `max()`. Is the largest value the same as the average?

```python
late_days.max()
```
