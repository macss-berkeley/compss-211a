## Practice: contract before repair

```python
sample = pd.DataFrame({"value": [1, 2, 3], "group": ["a", "a", "b"]})

def generated_summary(frame):
    return frame.groupby("group")["value"].sum().reset_index(name="mean")

raw_output = generated_summary(sample)
contract_results = {
    "row_per_group": len(raw_output) == sample["group"].nunique(),
    "column_names": list(raw_output.columns) == ["group", "mean"],
    "mean_is_correct": raw_output.set_index("group").loc["a", "mean"] == 1.5,
}
display(raw_output)
display(pd.Series(contract_results))
```
