## 4. Inspect and commit the new script — minutes 65–80

Predict how Git will label the new file, then run:

~~~bash

~~~

**Check:** the status should show <code>?? scripts/inspect_data.py</code>. Predict whether a normal diff will show that untracked file, then run:

~~~bash

~~~

**Check:** it should show no file contents because a normal <code>git diff</code> omits untracked files. Stage only the new script:

~~~bash

~~~

Confirm which file is staged:

~~~bash

~~~

Inspect the staged contents:

~~~bash

~~~

**Check:** read all four lines in the diff before committing them:

~~~bash

~~~

Verify the history and working tree one command at a time:

~~~bash

~~~

~~~bash

~~~

**Checkpoint:** the log should show two commits and the final status should be clean.

## 5. Make and safely recover one tracked-file mistake — minutes 80–95

Open <code>docs/notes.md</code>, add the line <code>Unverified note for recovery practice.</code>, and save it. Predict the status output, then run:

~~~bash
git status --short
~~~

**Check:** Git should mark only <code>docs/notes.md</code> as modified. Inspect that file's exact change:

~~~bash
git diff -- docs/notes.md
~~~

Read the exact changed line before restoring anything. Then restore only that file and verify the result:

~~~bash
git restore -- docs/notes.md
~~~

Verify directly that the file now matches the last commit:

~~~bash
git diff -- docs/notes.md
~~~

**Check:** the diff should print nothing. Reopen <code>docs/notes.md</code> and confirm that the unverified line is gone. Then check the repository and its history:

~~~bash
git status --short
~~~

~~~bash
git log --oneline -2
~~~

**Checkpoint:** the status should be clean, the unverified line should be gone, and both commits should remain. Do not use <code>reset --hard</code>.

### Checkpoint 5 discussion
