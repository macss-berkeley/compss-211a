## Practice: inspect a safe sandbox

```python
import shutil
import subprocess
```

```python
sandbox = GENERATED_DIR / "lab04_git_sandbox"
if sandbox.exists():
    shutil.rmtree(sandbox)
sandbox.mkdir()
subprocess.run(["git", "init", "-q"], cwd=sandbox, check=True)
subprocess.run(["git", "config", "user.email", "lab@example.invalid"], cwd=sandbox, check=True)
subprocess.run(["git", "config", "user.name", "COMPSS Lab"], cwd=sandbox, check=True)
note = sandbox / "field-note.txt"
note.write_text("verified source\n", encoding="utf-8")
subprocess.run(["git", "add", "field-note.txt"], cwd=sandbox, check=True)
subprocess.run(["git", "commit", "-q", "-m", "Add verified field note"], cwd=sandbox, check=True)
note.write_text("unverified overwrite\n", encoding="utf-8")
status_before = subprocess.run(
    ["git", "status", "--short"], cwd=sandbox, check=True,
    capture_output=True, text=True
).stdout.strip()
subprocess.run(["git", "restore", "field-note.txt"], cwd=sandbox, check=True)
status_after = subprocess.run(
    ["git", "status", "--short"], cwd=sandbox, check=True,
    capture_output=True, text=True
).stdout.strip()
print({"before": status_before, "after": status_after})
```
