````python
from pathlib import Path
import sys

EXPECTED_PYTHON = (3, 13, 15)
EXPECTED_NUMPY = "2.1.3"
EXPECTED_PANDAS = "2.2.3"


def find_course_root():
    """Find the course repository from the current notebook location."""
    for folder in (Path.cwd(), *Path.cwd().parents):
        if (folder / "pyproject.toml").exists() and (folder / "data").is_dir():
            return folder
    return None


checks = []


def record_check(label, passed, help_text):
    checks.append((label, bool(passed), help_text))


course_root = find_course_root()
record_check(
    "Course repository found",
    course_root is not None,
    "Open this notebook from the cloned course repository.",
)
record_check(
    "uv.lock found",
    course_root is not None and (course_root / "uv.lock").exists(),
    "Run `git pull`, then check that uv.lock is in the repository.",
)
record_check(
    "Course .venv is active",
    Path(sys.prefix).name == ".venv",
    f"Current environment: {sys.prefix}. Close Jupyter and restart it with `uv run jupyter lab`.",
)

python_version = sys.version_info[:3]
record_check(
    f"Python {'.'.join(map(str, python_version))}",
    python_version == EXPECTED_PYTHON,
    f"Expected Python {'.'.join(map(str, EXPECTED_PYTHON))}. Run `uv sync --frozen` and restart Jupyter.",
)

np = None
pd = None
try:
    import numpy as np
    import pandas as pd

    package_versions_ok = (
        np.__version__ == EXPECTED_NUMPY
        and pd.__version__ == EXPECTED_PANDAS
    )
    package_help = (
        f"Found NumPy {np.__version__} and pandas {pd.__version__}; "
        f"expected NumPy {EXPECTED_NUMPY} and pandas {EXPECTED_PANDAS}. "
        "Run `uv sync --frozen` and restart Jupyter."
    )
except Exception as error:
    package_versions_ok = False
    package_help = f"Package import failed: {type(error).__name__}: {error}"

record_check("NumPy and pandas imported", package_versions_ok, package_help)

try:
    data_path = course_root / "data" / "hw1_after_dark_commute_survey.csv"
    sample = pd.read_csv(data_path)
    data_ok = not sample.empty
    data_help = f"The file opened but contained no rows: {data_path}"
except Exception as error:
    data_ok = False
    data_help = f"Could not open the course data file: {type(error).__name__}: {error}"

record_check("Course data file opened", data_ok, data_help)

print("COMPSS 211A installation check\n")
for label, passed, _ in checks:
    symbol = "✓" if passed else "✗"
    print(f"{symbol} {label}")

failed = [(label, help_text) for label, passed, help_text in checks if not passed]
if not failed:
    print("\nREADY FOR WEEK 1")
else:
    print("\nNEEDS HELP")
    print("\nStart with the first failed check:")
    print(f"{failed[0][0]}: {failed[0][1]}")
    print("\nShow this message to an instructor or GSI. Do not reinstall everything.")
````
